### Data Visualization

```python
sns.barplot(x=df['Disease'].value_counts().index, y=df['Disease'].value_counts().values)
plt.title('Distribution of Disease')
plt.xlabel('Disease')
plt.ylabel('Count')
plt.show()
```

```python
sns.boxplot(x=df['Disease'], y=df['Age'])

plt.title('Boxplot of Age vs Disease')
plt.xlabel('Disease')       
plt.ylabel('Age')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxplot(x=df['Disease'], y=df['BMI'])

plt.title('Boxplot of BMI vs Disease')
plt.xlabel('Disease')       
plt.ylabel('BMI')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxplot(x=df['Disease'], y=df['GlucoseLevel'])

plt.title('Boxplot of Glucose Level vs Disease')
plt.xlabel('Disease')       
plt.ylabel('Glucose Level')
plt.xticks(rotation=45)
plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

### Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

```

```python
categorical_columns = df.select_dtypes(include='object').columns
```

```python
categorical_columns
```

```python
df[categorical_columns] = df[categorical_columns].apply(le.fit_transform)
```
